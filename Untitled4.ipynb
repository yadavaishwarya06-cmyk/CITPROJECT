```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = datasets.mnist.load_data()

# Normalize the images
x_train = x_train / 255.0
x_test = x_test / 255.0

# Add channel dimension
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# Create CNN model
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu',
                  input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model
model.fit(x_train, y_train, epochs=5)

# Test the model
test_loss, test_accuracy = model.evaluate(x_test, y_test)

print("Test Accuracy:", test_accuracy)
```

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# Hyperparameters
learning_rate = 0.001
batch_size = 32
epochs = 10
filters = 32
kernel_size = (3, 3)

# CNN Model
model = Sequential([
    Conv2D(filters, kernel_size, activation='relu',
           input_shape=(64, 64, 3)),
    MaxPooling2D(pool_size=(2, 2)),
    Flatten(),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

# Compile model
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Display model parameters
model.summary()

print("Learning Rate:", learning_rate)
print("Batch Size:", batch_size)
print("Epochs:", epochs)
print("Filters:", filters)
print("Kernel Size:", kernel_size)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# Load data
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize
x_train = x_train / 255.0
x_test = x_test / 255.0

# Convert labels
y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)

# RNN model
model = Sequential([
    SimpleRNN(64, input_shape=(28, 28)),
    Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Train
history = model.fit(
    x_train, y_train_cat,
    epochs=5,
    validation_data=(x_test, y_test_cat)
)

# Prediction
y_prob = model.predict(x_test)
y_pred = np.argmax(y_prob, axis=1)

# 1. Classification Report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# 2. Accuracy and Loss
loss, accuracy = model.evaluate(x_test, y_test_cat)
print("Accuracy:", accuracy)
print("Loss:", loss)

# 3. Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:")
print(cm)

plt.imshow(cm)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.colorbar()
plt.show()

# 4. ROC-AUC
auc = roc_auc_score(y_test_cat, y_prob, multi_class='ovr')
print("ROC-AUC:", auc)

# 5. Accuracy and Loss graph
plt.plot(history.history['accuracy'], label='Accuracy')
plt.plot(history.history['loss'], label='Loss')
plt.legend()
plt.title("Accuracy and Loss")
plt.show()
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# Training data
X = np.array([
    [1, 2, 3, 0],
    [2, 3, 4, 0],
    [4, 5, 6, 0],
    [5, 6, 7, 0]
])

y = np.array([0, 0, 1, 1])

# Create LSTM model
model = Sequential([
    Embedding(input_dim=10, output_dim=8, input_length=4),
    LSTM(16),
    Dense(1, activation='sigmoid')
])

# Compile
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train
model.fit(X, y, epochs=10, verbose=1)

# Prediction
prediction = model.predict(X)

print("Predictions:")
print(prediction)
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# Text data converted into numbers
X = np.array([
    [1, 2, 3, 0],
    [2, 3, 4, 0],
    [4, 5, 6, 0],
    [5, 6, 7, 0]
])

# Labels: 0 = Negative, 1 = Positive
y = np.array([0, 0, 1, 1])

# LSTM RNN model
model = Sequential()

model.add(Embedding(input_dim=10, output_dim=8, input_length=4))
model.add(LSTM(16))
model.add(Dense(1, activation='sigmoid'))

# Compile model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train model
model.fit(X, y, epochs=10)

# Prediction
prediction = model.predict(X)

print("Prediction:")
print(prediction)
```
